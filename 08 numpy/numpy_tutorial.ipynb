```python
import numpy as np
import time
```

```python
# execution time
size = 100_000_000

# python list computation
py_list = list(range(size))
print(type(py_list))
start_time = time.time()
sq_list = [x**2 for x in py_list]
end_time = time.time()
print(f"time consumed by python list {end_time-start_time} seconds")

# numpy arrays computation
np_arr = np.array(py_list)
print(type(np_arr))
start_time = time.time()
sq_arr = np_arr**2 # vectorization
end_time = time.time()
print(f"time consumed by numpy arrays {end_time-start_time} seconds")
```

```python
# memory
import sys

print(f"the size of python list= {sys.getsizeof(py_list) * len(py_list)} bytes")

print(f"the size of numpy array= {np_arr.nbytes} bytes")
```
