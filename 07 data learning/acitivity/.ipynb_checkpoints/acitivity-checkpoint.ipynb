```python
import json
```

```python
# Func to load data

def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)
    return data
```

```python
data = load_data("store_data.json")
print(type(data))
print(data)
```

```python
def clean_data(data):
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
    cleaned_data = []
    unique_users = set()
    for user in data:

        # data inconsistency
        raw_rating = str(user["rating"]).strip().lower()
        if raw_rating in text_to_num:
            raw_rating = text_to_num[raw_rating]
        user["rating"] = raw_rating

        # handling missing values
        raw_age = user.get(
            "age"
        )  # .get() function returns the value if exist else returns "None".
        if raw_age == None:
            user["age"] = None

        # deduplication
        if user["name"].strip() in unique_users:
            continue
        unique_users.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data


data = clean_data(data)
```

```python
print(data)
```

```python
# Getting meaningful insights


def get_insights(data):

    # avg rating
    total_rating = 0
    for user in data:
        total_rating += float(user["rating"])
        avg_rating = total_rating / len(data)
    print(f"The average rating: {avg_rating}")

    # percentage of users with poor rating(rating <3)
    poor_rating = 0
    for user in data:
        if float(user["rating"]) < 3:
            poor_rating += 1
    print(f"% of users with poor rating: {(poor_rating/len(data))*100}")


get_insights(data)
```

```python
# recommendation system


def get_recommendations(data):
    recommendation = []
    for user in data:
        current_recom = {}
        current_recom["name"] = user["name"]

        if float(user["rating"]) >= 4:
            current_recom["brand"] = "apple"
        else:
            current_recom["brand"] = "samsung"

        recommendation.append(current_recom)
    return recommendation


get_recommendations(data)
```
